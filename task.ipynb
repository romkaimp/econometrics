```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pyreadstat
```

```python
def load_sav(path: str):
    """
    Загружает .sav файл, возвращает DataFrame и метаданные.
    
    pyreadstat возвращает:
      df     — данные
      meta   — метаданные: имена колонок, подписи (labels), 
               значения-метки (value labels), формат SPSS
    """
    df, meta = pyreadstat.read_sav(path)
    return df, meta


# --- пример ---
PATH = "data/r34i_os_92.sav"        # <-- замените на путь к файлу
df, meta = load_sav(PATH)

print(f"Размер датасета: {df.shape[0]} строк × {df.shape[1]} столбцов\n")
print("Первые 5 строк:")
print(df.head())
```
